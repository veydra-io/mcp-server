```python
# Test 2: List initial repository files (should show initialized structure)
if test_model_id:
    print(f"Listing files in repository for model {test_model_id}...")
    
    response = requests.get(f"{BASE_URL}/v1/model/{test_model_id}/files", headers=headers)
    
    if response.status_code == 200:
        files_result = response.json()
        print("✅ Repository files retrieved successfully!")
        print(f"Total items: {files_result.get('total_items', 0)}")
        
        print("\n📁 Repository Structure:")
        for file_info in files_result.get('files', []):
            file_type = file_info.get('type', 'unknown')
            icon = "📁" if file_type == 'directory' else "📄"
            name = file_info.get('name', 'unknown')
            path = file_info.get('path', 'unknown')
            size = file_info.get('size', 'N/A')
            
            if file_type == 'directory':
                print(f"  {icon} {path}/")
            else:
                print(f"  {icon} {path} ({size} bytes)")
    else:
        print(f"❌ Failed to list files: {response.status_code}")
        print(f"Error: {response.text}")
else:
    print("⚠️ Skipping file listing - no model ID available")
```
